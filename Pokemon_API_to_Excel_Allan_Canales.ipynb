## 2 — Choose five Pokémon
You can change the names later.

```python
pokemon_names = ["pikachu", "bulbasaur", "charmander", "squirtle", "eevee"]
print(pokemon_names)
```

## 3 — Make one API request
`requests.get()` makes an HTTP request. `.json()` converts the JSON response into Python objects. Explore nested `types`.

```python
import requests
import pandas as pd

# First, inspect ONE response to understand the JSON.
url = "https://pokeapi.co/api/v2/pokemon/pikachu"
response = requests.get(url, timeout=20)
response.raise_for_status()  # Error if the request failed.
data = response.json()        # JSON → Python dictionary

print("Name:", data["name"])
print("Height (dm):", data["height"])
print("Types:", data["types"])
```

## 4 — Collect a small dataset
A list of dictionaries makes it easy to create a DataFrame. Notice how we extract the type name from nested JSON.

```python
# Collect the same four fields for each Pokémon.
rows = []

for name in pokemon_names:
    url = f"https://pokeapi.co/api/v2/pokemon/{name}"
    response = requests.get(url, timeout=20)
    response.raise_for_status()
    pokemon = response.json()

    row = {
        "name": pokemon["name"],
        "height_dm": pokemon["height"],
        "weight_hg": pokemon["weight"],
        "types": ", ".join(item["type"]["name"] for item in pokemon["types"]),
    }
    rows.append(row)

print(rows[0])
```

## 5 — Convert to a DataFrame

```python
# List of dictionaries → pandas DataFrame
df = pd.DataFrame(rows)
display(df)
```

## 6 — Mini exercise: transform the data
Convert `height_dm` to metres and `weight_hg` to kilograms, creating **two new columns**.

**Hint:** divide each original column by 10.

```python
# Exercise: convert measurements to friendlier units.
# 1 dm = 0.1 m; 1 hg = 0.1 kg
df["height_m"] = df["height_dm"] / 10
df["weight_kg"] = df["weight_hg"] / 10

display(df)
```

**Note:** PokéAPI gives height in decimetres and weight in hectograms, so dividing by 10 gives metres and kilograms. For example Pikachu is 4 dm, which is 0.4 m, and 60 hg, which is 6.0 kg.

## 7 — Save as Excel

```python
# Save the DataFrame as an Excel file in the current working directory.
filename = "pokemon_data.xlsx"
df.to_excel(filename, index=False)
print("Saved:", filename)

# The workflow diagram also asks for a CSV, so we save one too.
df.to_csv("pokemon_data.csv", index=False)
print("Saved: pokemon_data.csv")
```
